## Part A - Numpy Tasks

### Q1 - Student Marks Analysis using numpy arrays
• Mean  
• Median  
• Maximum value  
• Minimum value  
• Standard deviation  

```python
import numpy as np
marks = np.array([78, 85, 92, 60, 73, 88, 95, 67, 81, 74])
print("Marks of 10 students:", marks)
print()
print("Mean:", np.mean(marks))
print("Median:", np.median(marks))
print("Maximum value:", np.max(marks))
print("Minimum value:", np.min(marks))
print("Standard deviation:", np.std(marks))
```

### Q2 - Reshape, Shape & Dimensions
Create a NumPy array from 1 to 20 and:  
• Reshape it into a 4×5 matrix  
• Find shape of array  
• Find dimensions of array  

```python
arr = np.arange(1, 21)
print("Original array:", arr)
matrix = arr.reshape(4, 5)
print("\nReshaped into 4*5 matrix:\n", matrix)
print("\nShape of array:", matrix.shape)
print("\nDimensions of array:", matrix.ndim)
```

```python
print("First 5 elements:  ", arr[:5])
print("Last 5 elements:   ", arr[-5:])
print("Alternate elements:", arr[::2])
```

## Part - B Pandas Tasks

### Q5 - Load the dataset

```python
import pandas as pd

df = pd.read_csv("titanic.csv")
```

### Q6 - Display Basic Info
• First 5 rows  
• Dataset shape  
• Column names  
• Dataset information  

```python
# First 5 rows
print("First 5 rows:")
print(df.head())

# Dataset shape
print("\nDataset shape (rows, columns):", df.shape)

# Column names
print("\nColumn names:", df.columns.tolist()), 

# Dataset information
print("\nDataset information:")
df.info()
```

### Q7 - Passenger Statistics
• Total number of passengers   
• Total male and female passengers   
• Average age   
• Maximum fare   
• Minimum fare   

```python
# Total number of passengers
print("Total number of passengers:", len(df))

# Total male and female passengers
print("\nTotal male and female passengers:", df["Sex"].value_counts())

# Average age
print("\nAverage age:", df["Age"].mean().round(2))

# Maximum fare
print("Maximum fare:", df["Fare"].max())

# Minimum fare
print("Minimum fare:", df["Fare"].min())
```

### Q8 - Filtering Operations
• Passengers older than 30    
• Female passengers   
• Passengers who survived   
• First-class passengers

```python
# Passengers older than 30
older_30 = df[df['Age'] > 30]
print("Passengers older than 30:", len(older_30))
print(older_30.head())

# Female passengers
female = df[df['Sex'] == 'female']
print("\nFemale passengers:", len(female))
print(female.head())

# Passengers who survived (1 = survived)
survived = df[df['Survived'] == 1]
print("\nPassengers who survived:", len(survived))
print(survived.head())

# First-class passengers (Pclass = 1)
firstclass = df[df['Pclass'] == 1]
print("\nFirst-class passengers:", len(firstclass))
print(firstclass.head())
```

### Q9 - Check Missing Values

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

### Q10 - Handle Missing Values
* Fill missing Age values with average age
* Drop rows with missing Embarked values

```python
# Fill missing Age values with average age
avg_age = df['Age'].mean()
df['Age'] = df['Age'].fillna(avg_age)
print("Missing Age values filled with average age:", round(avg_age, 2))

# Drop rows with missing Embarked values
df = df.dropna(subset = ['Embarked'])
print("Rows with missing embarked values dropped.")

# Verify no missing values remain in these columns
print("\nMissing values after handling:")
print(df[['Age', 'Embarked']].isnull().sum())
```

## Part C - Matplotlib Tasks

### Q11 - Bar Chart: Passenger Count by Class

```python
import matplotlib.pyplot as plt

class_counts = df['Pclass'].value_counts().sort_index()

plt.figure(figsize = (6, 4))
plt.bar(['1st Class', '2nd Class', '3rd Class'], class_counts.values, color = ['gold', 'skyblue', 'salmon'])
plt.title("Passenger Count by Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.tight_layout()
plt.show()
```

### Q12 - Pie Chart: Male vs Female Passengers

```python
gender_counts = df['Sex'].value_counts()

plt.figure(figsize = (6, 4))
plt.pie(gender_counts.values, labels = gender_counts.index, autopct = '%1.1f%%', colors = ['skyblue', 'lightpink'])
plt.title("Male vs Female Passengers")
plt.tight_layout()
plt.show()
```

### Q13 - Histogram: Age Distribution

```python
plt.figure(figsize = (7, 4))
plt.hist(df['Age'], bins = 20, color = 'steelblue', edgecolor = 'black')
plt.title("Age Distribution of Passengers")
plt.xlabel('Age')
plt.ylabel('Number of Passengers')
plt.tight_layout()
plt.show()
```

### Q14 - Scatter Plot: Age vs Fare

```python
plt.figure(figsize = (7, 4))
plt.scatter(df['Age'], df['Fare'], color = 'purple', alpha = 0.5)
plt.title("Age vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.tight_layout()
plt.show()
```
