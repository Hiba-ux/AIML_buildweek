# Salary Prediction - Inference Notebook

This notebook loads the saved pipeline and predicts salaries for new employee data.

> **Run `train_pipeline.ipynb` first** to generate `salary_pipeline.pkl`

## 1. Import Libraries & Load Pipeline

```python
import pandas as pd
import numpy as np
import joblib

# Load the saved pipeline
pipeline = joblib.load('salary_pipeline.pkl')
print('Pipeline loaded successfully!')
print(f'Model type: {type(pipeline.named_steps["model"]).__name__}')
```

## 2. Define New Employee Data

```python
# Define new employees to predict salaries for
new_employees = pd.DataFrame([
    {
        'Age': 29,
        'Gender': 'Male',
        'Education': "Bachelor's",
        'Job_Title': 'Software Engineer',
        'Years_Experience': 4,
        'Department': 'Engineering'
    },
    {
        'Age': 35,
        'Gender': 'Female',
        'Education': "Master's",
        'Job_Title': 'Data Scientist',
        'Years_Experience': 8,
        'Department': 'Data Science'
    },
    {
        'Age': 45,
        'Gender': 'Male',
        'Education': 'PhD',
        'Job_Title': 'Director',
        'Years_Experience': 18,
        'Department': 'Management'
    },
    {
        'Age': 27,
        'Gender': 'Female',
        'Education': "Bachelor's",
        'Job_Title': 'HR Specialist',
        'Years_Experience': 2,
        'Department': 'HR'
    },
    {
        'Age': 40,
        'Gender': 'Male',
        'Education': "Master's",
        'Job_Title': 'Product Manager',
        'Years_Experience': 12,
        'Department': 'Product'
    }
])

print('New Employee Data:')
new_employees
```

## 3. Predict Salaries

```python
# Predict using the loaded pipeline
predicted_salaries = pipeline.predict(new_employees)

# Display results
results = new_employees.copy()
results['Predicted_Salary'] = predicted_salaries.round(2)

print('Salary Predictions:')
print('=' * 70)
for i, row in results.iterrows():
    print(f"  {row['Job_Title']:30s} | {row['Education']:12s} | {row['Years_Experience']} yrs exp | ${row['Predicted_Salary']:>10,.2f}")
print('=' * 70)
```

## 4. Interactive Single Prediction

```python
# Predict salary for a single custom employee
# Modify these values to predict for any employee!

custom_employee = pd.DataFrame([{
    'Age': 33,                         # Age in years
    'Gender': 'Female',                # Male / Female
    'Education': "Master's",           # Bachelor's / Master's / PhD
    'Job_Title': 'ML Engineer',        # Job title
    'Years_Experience': 7,             # Years of experience
    'Department': 'Engineering'        # Engineering / Data Science / HR / etc.
}])

salary_prediction = pipeline.predict(custom_employee)[0]

print('Custom Salary Prediction')
print('=' * 40)
print(f"  Name       : [Custom Employee]")
print(f"  Age        : {custom_employee['Age'].values[0]}")
print(f"  Gender     : {custom_employee['Gender'].values[0]}")
print(f"  Education  : {custom_employee['Education'].values[0]}")
print(f"  Job Title  : {custom_employee['Job_Title'].values[0]}")
print(f"  Experience : {custom_employee['Years_Experience'].values[0]} years")
print(f"  Department : {custom_employee['Department'].values[0]}")
print('=' * 40)
print(f"  Predicted Salary: ${salary_prediction:,.2f}")
```
